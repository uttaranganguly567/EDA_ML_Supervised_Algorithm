```python
import sys
class Employee:
    def show(self):
        print(f"The python version is {sys.version}")
    def showemp(self, name: str, address: str, phone: int):
        self.name = name
        self.address = address
        self.phone = phone
        print(f"The name is {self.name}, address is {self.address} and phone is {self.phone}")
```

```python
e1 = Employee() #The object is stored in the heap memory / Dynamic memory
# 1. Non Garbage collectable heap.
# 2. Garbage collectable heap.
# C++ and uses non garbage collectable heap hence we have to manually delete the unused memory by free() function.
# Java and Python uses garbage collectable heap hence we have no concern of the unused memory.
# It is taken care by PVM and JVM by using a low-priority background DAEMON THREAD called Gargabe Collector.
```

```python
reg = LinearRegression() #Object of Linear Regression
reg
```

```python
# Training the model -
X = df[['area']]
y = df['price']
```

```python
X
```

```python
y
```

```python
reg.fit(X, y)
```

```python
#Lets predict for a new VALUE
reg.predict([[7000]])
```

```python
print(reg.predict([[7000]]))
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
y = (reg.coef_) * 7000 + (reg.intercept_)
```

```python
y
```

```python
def brain(model: LinearRegression, sqft: int):
    return model.coef_ * sqft + model.intercept_
```

```python
brain(reg, 7000)
```

```python
reg.predict([[5000]])
```

```python
#FIND ACCURACY OF THE MODEL
from sklearn.metrics import r2_score
y_original = df.price
y_predicted = reg.predict(df[['area']])
r = r2_score(y_original, y_predicted)
print(f"R squared value {r}")
```
